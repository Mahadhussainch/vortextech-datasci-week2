### Objective:
To explore numeric and categorical relationships within the Supermarket Sales dataset and extract three specific, meaningful insights backed by statistical visualizations.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the supermarket sales dataset
df = pd.read_csv('SMA.csv')

```

### Step 1: Checking and removing nulls and duplicates

```python

df_clean = df.dropna().drop_duplicates()
print(f"Original shape: {df.shape} | Cleaned shape: {df_clean.shape}")
```

### Step 2: Numeric Correlation Heatmap
Identifying the strongest positive and negative relationships across all numeric variables to guide the scatter plot analysis.

```python
# Step 2: Generate and visualize correlation matrix
numeric_cols = df_clean.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(10, 8))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numeric Variables')
plt.show()
```

### Step 3: Driving the Insights

### Insight 1: Unit Price vs. Sales
*Observation: There is a strong positive correlation between the unit price of an item and the total sales value. This suggests that higher-priced items significantly drive overall revenue per transaction, despite potential variations in quantity purchased.*

```python
# Step 3 & 4: Visualizing the first strong relationship
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='Unit price', y='Sales', alpha=0.6)
plt.title('Relationship: Unit Price vs. Total Sales')
plt.show()
```

### Insight 2: COGS vs. Gross Income
*Observation: The Cost of Goods Sold (COGS) has a perfect positive relationship with Gross Income. Since the gross margin percentage is fixed in this dataset (4.76%), the gross income scales linearly and predictably with the cost of goods sold.*

```python
# Step 4: Visualizing the second strong relationship
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='cogs', y='gross income', color='orange', alpha=0.6)
plt.title('Relationship: Cost of Goods Sold vs. Gross Income')
plt.show()
```

### Insight 3: Average Sales by Product Line
*Observation: When grouping by product line, we can see notable variations in average transaction values. (Review the generated chart to identify which specific product line yields the highest average sales).*

```python
# Step 5: Categorical grouping and bar chart
cat_analysis = df_clean.groupby('Product line')['Sales'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(data=cat_analysis, x='Product line', y='Sales', palette='viridis')
plt.title('Average Sales by Product Line')
plt.ylabel('Average Sales')
plt.xlabel('Product Line')
plt.xticks(rotation=45)
plt.show()
```

### Conclusion
1. **Pricing Drives Revenue:** Unit price strongly influences the total sales amount per invoice.
2. **Fixed Margins:** Gross income scales perfectly linearly with COGS due to a static gross margin percentage across all transactions.
3. **Product Line Variance:** Certain product categories consistently yield higher average sales per transaction, highlighting key revenue drivers.
